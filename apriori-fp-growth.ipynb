```python
import pandas as pd

df = pd.read_excel('supermarket.xlsx', sheet_name='Sayfa2')

df.head()
```

```python
import warnings
warnings.filterwarnings("ignore")

# sadece 0/1 yap
df = df.applymap(lambda x: 1 if x == 1 else 0)

# bool'a çevir
df = df.astype(bool)

print(df.shape)
```

```python
# Apriori parametre ayarı
min_support = 0.15
min_conf = 0.5
```

```python
#apriori

from mlxtend.frequent_patterns import apriori, association_rules

frequent_apriori = apriori(df, min_support=min_support, use_colnames=True)

rules_apriori = association_rules(frequent_apriori, metric="confidence", min_threshold=min_conf)

print("Apriori rule count:", len(rules_apriori))

rules_apriori.sort_values(by="confidence", ascending=False).head(10)
```

```python
from mlxtend.frequent_patterns import fpgrowth

frequent_fp = fpgrowth(df, min_support=min_support, use_colnames=True)

rules_fp = association_rules(frequent_fp, metric="confidence", min_threshold=min_conf)

print("FP-Growth rule count:", len(rules_fp))

rules_fp.sort_values(by="confidence", ascending=False).head(10)
```

```python
import time

#performans karşılaştırma
start = time.time()
apriori(df, min_support=min_support, use_colnames=True)
print("Apriori time:", time.time() - start)

start = time.time()
fpgrowth(df, min_support=min_support, use_colnames=True)
print("FP-Growth time:", time.time() - start)
```

```python
#ortak kurallar
set_ap = set(rules_apriori['antecedents'].astype(str) + "->" + rules_apriori['consequents'].astype(str))
set_fp = set(rules_fp['antecedents'].astype(str) + "->" + rules_fp['consequents'].astype(str))

print("Common rules:", len(set_ap & set_fp))
```
